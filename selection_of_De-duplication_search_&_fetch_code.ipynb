```python
# ============================================
# Bee Toxicogenomics Paper Classifier
# For Google Colab
# ============================================

import pandas as pd
import re
import numpy as np
from google.colab import files

# Upload your Excel file
print("Please upload your Excel file (Reviewd.Deduplicated_Bee_Toxicogenomics_Master_2010_2026.xlsx)")
uploaded = files.upload()

# Get the filename
file_name = list(uploaded.keys())[0]

# Read the Excel file
df = pd.read_excel(file_name, sheet_name='Sheet1')

print(f"\nLoaded {len(df)} papers from {file_name}")
print(f"Columns: {df.columns.tolist()}")
print(f"\nFirst 5 rows:")
print(df.head())

# ============================================
# Define Keywords for Classification
# ============================================

# Keywords for "molecular/omics" methods (Selected)
MOLECULAR_KEYWORDS = [
    'transcriptom', 'rna-seq', 'rnaseq', 'rna seq', 'microarray',
    'proteom', 'mass spec', 'lc-ms', 'gc-ms',
    'metabolom', 'nmr',
    'genom', 'whole genome', 'genome-wide',
    'multi-omics', 'multiomics',
    'gene expression', 'differential expression',
    'qPCR', 'rt-qpcr', 'real-time pcr', 'quantitative pcr',
    'knockdown', 'rnai', 'interference rna',
    'crispr', 'cas9',
    'epigenetic', 'methylation',
    'pathway', 'kegg', 'go enrichment'
]

# Keywords for "bee" topics
BEE_KEYWORDS = [
    'honey bee', 'honeybee', 'apis mellifera', 'apis cerana',
    'bumble bee', 'bumblebee', 'bombus',
    'stingless bee', 'melipona', 'tetragonisca',
    'bee', 'bees', 'pollinator', 'hive', 'colony'
]

# Keywords for "pesticide" topics
PESTICIDE_KEYWORDS = [
    'pesticide', 'insecticide', 'fungicide', 'herbicide', 'acaricide',
    'neonicotinoid', 'imidacloprid', 'thiamethoxam', 'clothianidin',
    'flupyradifurone', 'sulfoxaflor', 'spinosad',
    'glyphosate', 'paraquat', 'atrazine',
    'pyrethroid', 'deltamethrin', 'permethrin',
    'organophosphate', 'malathion', 'chlorpyrifos',
    'varroa', 'amitraz', 'fluvalinate', 'coumaphos'
]

# Keywords that suggest the paper is NOT relevant (reviews, non-bee, methods-only)
EXCLUDE_KEYWORDS = [
    'review', 'systematic review', 'meta-analysis',
    'perspective', 'opinion', 'commentary',
    'editorial', 'letter', 'correction',
    'plant', 'crop', 'agricultural', 'soil',
    'zebrafish', 'mouse', 'rat', 'human', 'cell line',
    'method validation', 'analytical method', 'residue analysis'
]

# ============================================
# Classification Functions
# ============================================

def check_keywords(text, keyword_list):
    """Check if any keyword in list appears in text (case-insensitive)"""
    if pd.isna(text) or text == '':
        return False
    text_lower = str(text).lower()
    for keyword in keyword_list:
        if keyword.lower() in text_lower:
            return True
    return False

def count_keywords(text, keyword_list):
    """Count how many keywords appear in text"""
    if pd.isna(text) or text == '':
        return 0
    text_lower = str(text).lower()
    count = 0
    for keyword in keyword_list:
        if keyword.lower() in text_lower:
            count += 1
    return count

def classify_paper(row):
    """
    Classify a paper as 'Selected' or 'Remove' based on title and abstract.
    """
    # Combine title and abstract for analysis
    title = str(row.get('Title', '')) if pd.notna(row.get('Title', '')) else ''
    abstract = str(row.get('Abstract', '')) if pd.notna(row.get('Abstract', '')) else ''
    combined = title + ' ' + abstract

    # Check if it's a review or irrelevant
    if check_keywords(combined, EXCLUDE_KEYWORDS):
        return 'Remove', 'Review or non-research paper'

    # Check if it's about bees
    if not check_keywords(combined, BEE_KEYWORDS):
        return 'Remove', 'Not about bees/pollinators'

    # Check if it's about pesticides
    if not check_keywords(combined, PESTICIDE_KEYWORDS):
        return 'Remove', 'Not about pesticides'

    # Count molecular keywords
    molecular_count = count_keywords(combined, MOLECULAR_KEYWORDS)

    # Special case: RT-PCR only (with no other omics)
    rt_pcr_only = False
    if 'rt-pcr' in combined.lower() or 'rt-qpcr' in combined.lower() or 'qpcr' in combined.lower():
        other_omics = ['transcriptom', 'proteom', 'metabolom', 'genom', 'rna-seq', 'microarray', 'multi-omics']
        has_other_omics = any(kw in combined.lower() for kw in other_omics)
        if not has_other_omics:
            rt_pcr_only = True

    # Decision logic
    if molecular_count >= 2 or ('transcriptom' in combined.lower() and 'gene expression' in combined.lower()):
        return 'Selected', f'Omics/molecular study (count: {molecular_count})'
    elif rt_pcr_only:
        return 'Remove', 'RT-PCR only (no other omics)'
    elif molecular_count == 1:
        return 'Selected', f'Some molecular component (count: 1)'
    else:
        return 'Remove', 'No molecular/omics methods detected'

# ============================================
# Apply Classification
# ============================================

# Create new columns
df['Auto_Status'] = ''
df['Auto_Reason'] = ''

for idx, row in df.iterrows():
    status, reason = classify_paper(row)
    df.at[idx, 'Auto_Status'] = status
    df.at[idx, 'Auto_Reason'] = reason

# ============================================
# Show Results Summary
# ============================================

print("\n" + "="*80)
print("CLASSIFICATION RESULTS SUMMARY")
print("="*80)

status_counts = df['Auto_Status'].value_counts()
print(f"\nTotal papers: {len(df)}")
print(f"Selected: {status_counts.get('Selected', 0)}")
print(f"Remove: {status_counts.get('Remove', 0)}")

print("\n" + "-"*80)
print("DETAILED BREAKDOWN BY REASON")
print("-"*80)

reason_counts = df['Auto_Reason'].value_counts()
for reason, count in reason_counts.items():
    print(f"  {reason}: {count}")

# ============================================
# Preview Selected Papers
# ============================================

print("\n" + "-"*80)
print("SELECTED PAPERS (First 20)")
print("-"*80)

selected_df = df[df['Auto_Status'] == 'Selected']
selected_columns = ['Title', 'Auto_Status', 'Auto_Reason']
display_columns = [col for col in selected_columns if col in df.columns]

if len(selected_df) > 0:
    print(selected_df[display_columns].head(20).to_string(index=False))
else:
    print("No papers selected.")

# ============================================
# Preview Removed Papers
# ============================================

print("\n" + "-"*80)
print("REMOVED PAPERS (First 20)")
print("-"*80)

removed_df = df[df['Auto_Status'] == 'Remove']
if len(removed_df) > 0:
    print(removed_df[display_columns].head(20).to_string(index=False))
else:
    print("No papers removed.")

# ============================================
# Update the "Status" Column (if it exists)
# ============================================

# If you have a 'Status' column, update it with the classification
if 'Status' in df.columns:
    df['Status'] = df['Auto_Status']
    print("\nUpdated 'Status' column with classifications.")

# ============================================
# Save the Updated Excel File
# ============================================

# Create a new filename with "_classified" suffix
output_file = file_name.replace('.xlsx', '_classified.xlsx')
if output_file == file_name:
    output_file = 'classified_bee_papers.xlsx'

# Write to Excel with both sheets
with pd.ExcelWriter(output_file, engine='openpyxl') as writer:
    df.to_excel(writer, sheet_name='Sheet1', index=False)

    # Also save the summary as a separate sheet
    summary_data = {
        'Category': ['Total Papers', 'Selected', 'Removed'],
        'Count': [len(df), status_counts.get('Selected', 0), status_counts.get('Remove', 0)]
    }
    summary_df = pd.DataFrame(summary_data)
    summary_df.to_excel(writer, sheet_name='Summary', index=False)

print(f"\n✅ Updated file saved as: {output_file}")

# ============================================
# Download the File
# ============================================

files.download(output_file)

print("\n✅ File downloaded successfully!")
```
